```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, classification_report
```

# 데이터 준비

```python
df = sns.load_dataset("titanic")
```

```python
df.head()
```

# 데이터 탐색

```python
df.info()
```

```python
# 컬럼별 결측값의 개수
df.isna().sum()
```

```python
# 결측치가 많은 deck 열은 삭제, embarked 와 내용이 겹치는 embark_town 열을 삭제
rdf = df.drop(["deck", "embark_town"], axis = 1)
```

```python
rdf.columns
```

```python
# age 열에 결측치가 있는 모든 행을 삭제 (177개 행)
rdf = rdf.dropna(subset = ["age"], axis = 0)
```

```python
len(rdf)
```

```python
rdf["embarked"].value_counts()
```

```python
# embarked 열의 NaN 값을 승선도시 중에서 가장 많이 출현한 값으로 치환하기
most_freq = rdf["embarked"].value_counts().idxmax()
```

```python
most_freq
```

```python
rdf["embarked"] = rdf["embarked"].fillna(most_freq)
```

```python
rdf.isna().sum()
```

# 분석에 사용할 속성을 선택

```python
rdf.head()
```

```python
ndf = rdf[["survived", "pclass", "sex", "age", "sibsp", "parch", "embarked"]]
```

```python
ndf.head()
```

```python
# 원핫인코딩 - 범주형 데이터를 인식할 수 있도록 숫자형으로 변환
onehot_sex = pd.get_dummies(ndf["sex"], dtype = int)
```

```python
onehot_sex
```

```python
ndf = pd.concat([ndf, onehot_sex], axis = 1)
```

```python
ndf.head()
```

```python
onehot_embarked = pd.get_dummies(ndf["embarked"], prefix = "town", dtype = int)
```

```python
onehot_embarked.head()
```

```python
ndf = pd.concat([ndf, onehot_embarked], axis = 1)
```

```python
ndf.head()
```

```python
ndf = ndf.drop(["sex", "embarked"], axis = 1)
```

```python
ndf.head()
```

# 데이터셋 분할

```python
x = ndf.drop("survived", axis = 1) # 독립 변수 x
y = ndf["survived"] # 종속 변수 y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.3, stratify = y,
                                                    random_state = 26)
```

```python
x_train.shape, x_test.shape
```

# KNN 분류

```python
# KNN 분류 모형 객체 생성 (k = 5)
knn = KNeighborsClassifier(n_neighbors = 5)
```

```python
# 모델 학습
knn.fit(x_train, y_train)
```

```python
# test data 예측
y_pred = knn.predict(x_test)
```

```python
pd.DataFrame({"ans": y_test, "pred": y_pred})
```

```python
# 모형 성능 평가 - Confusion Matrix
knn_matrix = confusion_matrix(y_test, y_pred)
```

```python
# 0인 데이터를 0이라 한 경우, 0인 데이터를 1이라 한 경우
# 1인 데이터를 0이라 한 경우, 1인 데이터를 1이라 한 경우
print(knn_matrix)
```

```python
knn_report = classification_report(y_test, y_pred)
```

```python
print(knn_report)
```

- precision(정밀도)
    - 양성으로 예측한 데이터 중 실제로 양성인 데이터의 비율
 
- recall(재현율)
    - 실제 양성인 데이터 중 양성으로 예측한 비율
 
- 코로나 검사키트의 경우
    - 민감도 90% 이상
        - 실제 양성인 사람이 코로나 검사를 하면 양성으로 예측될 확률 90% 이상
    - 특이도 99% 이상
        - 실제 음성인 사람이 코로나 검사를 하면 음성으로 예측될 확률 99% 이상
