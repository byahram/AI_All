# 토크나이징

- 토큰(token)
    - 일정한 의미가 있는 가장 작은 단어들
 
- 토크나이징
    - 주어진 문장에서 토큰 단위로 정보를 나누는 작업
    - 문장 형태의 데이터를 처리하기 위해 제일 처음 수행하는 기본적인 작업
    - 텍스트 전처리 과정에서 사용됨
    - 토크나이징을 어떻게 하느냐에 따라 성능 차이가 날 수 있음
 
- 한국어 토크나이징에는 KoNLPy가 많이 사용됨

## Kkma

- 서울대학교 IDS(Intelligent Data Systems) 연구실에서 자연어 처리를 위해 개발한 한국어 형태소 분석기

- 다른 형태소 분석기에 비해 속도가 느리지만 지원하는 품사 태그가 다양해서 속도가 느리더라도 정확한 품사 정보가 필요할 때 사용

```python
from konlpy.tag import Kkma
```

```python
# 꼬꼬마 형태소 분석기 객체 생성
kkma = Kkma()
```

```python
text = "아버지가 방에 들어갑니다."

# 형태소 추출
morphs = kkma.morphs(text)
print(morphs)
```

- 형태소 단위로 토크나이징
- 리스트 형태로 반환됨

```python
# 형태소와 품사 태그 추출
pos = kkma.pos(text)
print(pos)
```

- 품사 태깅
- 리스트 형태로 반환됨

```python
# 명사만 추출
nouns = kkma.nouns(text)
print(nouns)
```

```python
# 문장 분리
sentences = "오늘 날씨는 어때요? 내일은 덥다던데."
s = kkma.sentences(sentences)
print(s)
```

## OKT

- Open-source Korean Text Processor

- 트위터에서 개발한 Twitter 한국어 처리기에서 파생된 오픈소스 한국어 처리기

- 간단한 한국어 처리를 통해 색인어를 추출하는 목표를 갖고 있기 때문에 완전한 수준의 형태소 분석을 지향하지 않음

```python
from konlpy.tag import Okt
```

```python
okt = Okt()
```

```python
# 형태소 추출
morphs = okt.morphs(text)
print(morphs)
```

```python
# 형태소와 품사 태그 추출
pos = okt.pos(text)
print(pos)
```

```python
# 명사만 추출
nouns = okt.nouns(text)
print(nouns)
```

```python
# 정규화, 어구 추출
text = "오늘 날씨가 좋아욬ㅋㅋ"
print(okt.normalize(text))
print(okt.phrases(text))
```

## Komoran

- Korean Morphological ANalyzer

- Shineware 에서 자바로 개발한 한국어 형태소 분석기

- 공백이 포함된 형태소 단위로도 분석이 가능

```python
from konlpy.tag import Komoran
```

```python
komoran = Komoran()
```

```python
morphs = komoran.morphs(text)
print(morphs)
```

```python
pos = komoran.pos(text)
print(pos)
```

```python
nouns = komoran.nouns(text)
print(nouns)
```

# 사용자 사전 구축

- 신조어는 형태소 분석기에서 인식이 안 되는 경우가 많음

- 이를 해결하기 위해 대부분의 형태소 분석기들은 사용자 사전을 추가할 수 있도록 구성됨

```python
text = "우리 챗봇은 엔엘피를 좋아해."
pos = komoran.pos(text)
print(pos)
```

```python
komoran = Komoran(userdic = "user_dic.tsv")
```

```python
pos = komoran.pos(text)
print(pos)
```
