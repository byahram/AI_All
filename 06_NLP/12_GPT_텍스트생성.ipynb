```python
import keras
from keras import layers
import keras_nlp
from keras_nlp.layers import ReversibleEmbedding
import tensorflow as tf
from transformers import pipeline, set_seed, AutoTokenizer, AutoModelForCausalLM, GenerationConfig
```

# 허깅페이스로 다양한 텍스트 생성

- transformers 라이브러리를 사용해 gpt-2 모델을 로드하고 사용

```python
set_seed(33)
hf_gpt1 = pipeline("text-generation", model = "openai-community/openai-gpt")
hf_gpt1("stay hungry, stay", max_new_tokens = 20, truncation = True)
```

```python
# 한 번에 여러 문장을 생성
set_seed(33)
hf_gpt1("stay hungry, stay", max_new_tokens = 20, truncation = True, num_return_sequences = 3)
```

```python
# AutoModel 활용
hf_gpt2_tokenizer = AutoTokenizer.from_pretrained("gpt2")
hf_gpt2 = AutoModelForCausalLM.from_pretrained("gpt2")
```

```python
# generate() 메서드는 텐서 입력을 받기 때문에 파이토치 텐서로 토큰값을 반환
prep_data = hf_gpt2_tokenizer("stay hungry, stay", return_tensors = "pt")
prep_data
```

```python
# generate() 에 토크나이저에서 반환된 값을 전달
set_seed(33)
outputs = hf_gpt2.generate(**prep_data, max_length = 20)
outputs
```

```python
# 토큰 아이디를 문자열로 변환
hf_gpt2_tokenizer.batch_decode(outputs)
```

- generate() 메서드는 기본적으로 그리디 샘플링을 수행함
