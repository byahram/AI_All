```python
# ---------------------------------------------------------
# ① 데이터 확인과 기본 통계
# ---------------------------------------------------------

import pandas as pd
import numpy as np

df = pd.read_csv('data.csv')

df.head()
df.shape
df.info()
df.describe()
df.isna().sum()

# 기본 통계
df['col'].mean()
df['col'].median()
df['col'].std()
df['col'].var()
df['col'].max()
df['col'].min()
df['col'].quantile(0.75)

# 최빈값
df['col'].mode()[0]

# 고유값 개수
df['col'].nunique()
```

```python
# ---------------------------------------------------------
# ② 조건 필터·정렬·순위
# ---------------------------------------------------------

# 조건 필터
df[df['age'] >= 30]

# 복합 조건
df[(df['age'] >= 30) & (df['score'] > 80)]

# 정렬
df.sort_values('score', ascending=False)

# 상위 5개
df.nlargest(5, 'score')

# 중복 제거
df.drop_duplicates(subset=['id'])

# 순위
df['rank'] = df['score'].rank(
    ascending=False,
    method='min'
)
```

```python
# ---------------------------------------------------------
# ③ 결측치 처리
# ---------------------------------------------------------

# 결측치 개수
df.isna().sum()

# 평균 대체
df['col'] = df['col'].fillna(df['col'].mean())

# 중앙값 대체
df['col'] = df['col'].fillna(df['col'].median())

# 그룹별 평균 대체
df['col'] = df['col'].fillna(
    df.groupby('group')['col'].transform('mean')
)

# 결측 행 제거
df = df.dropna(subset=['col'])
```

```python
# ---------------------------------------------------------
# ④ 이상치 — IQR / Z-score 
# ---------------------------------------------------------

# IQR
q1 = df['col'].quantile(0.25)
q3 = df['col'].quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outlier = df[
    (df['col'] < lower) |
    (df['col'] > upper)
]

print(len(outlier))

# Z-score 
from scipy.stats import zscore

z = zscore(df['col'].dropna())
print((abs(z) > 3).sum())
```

```python
# ---------------------------------------------------------
# ⑤ 그룹 집계·날짜·문자열
# ---------------------------------------------------------

# 그룹 평균
df.groupby('group')['score'].mean()

# 그룹별 합계
df.groupby('group')['sales'].sum()

# 다중 집계
df.groupby('group')['score'].agg(
    ['mean', 'sum', 'count']
)

# 날짜
df['date'] = pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day

# 날짜 차이
(df['end'] - df['start']).dt.days

# 문자열
df['name'].str.contains('김', na=False)
df['name'].str.len()
df['name'].str.split('-').str[0]
df['name'].str.replace('-', '', regex=False)
```

```python
# ---------------------------------------------------------
# ⑥ 정규화·표준화·상관계수
# ---------------------------------------------------------

# MinMax 정규화
x = df['col']
scaled = (x - x.min()) / (x.max() - x.min())

# 표준화
standard = (x - x.mean()) / x.std()

# 상관계수
df[['a', 'b']].corr()
df['a'].corr(df['b'])
```
