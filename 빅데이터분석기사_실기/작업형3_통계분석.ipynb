## 가설검정

```python
# ---------------------------------------------------------
# 가설검정 - 평균을 비교하는 문제
# ---------------------------------------------------------
from scipy import stats

# ---------------------------------------------------------
# 1. 단일표본 t검정
# 키워드: 한 집단, 기준값, 모집단 평균, 특정 수치와 비교
# ---------------------------------------------------------

# ------------ 단일표본 t검정 ------------
# 대립가설: 원두 상품의 평균 무게는 120g이 아니다.
stats.ttest_1samp(df['무게'], 120)
stats.ttest_1samp(df['무게'], 120, alternative='two-sided')

# 대립가설: 원두 상품의 평균 무게는 120g보다 크다.
stats.ttest_1samp(df['무게'], 120, alternative='greater')

# 대립가설: 원두 상품의 평균 무게는 120g보다 작다.
stats.ttest_1samp(df['무게'], 120, alternative='less')

# ------------ Shapiro-Wilk 정규성 검정 ------------
# 귀무가설: 정규분포를 따른다.
# 대립가설: 정규분포를 따르지 않는다.
stats.shapiro(df['무게'])

# ------------ Wilcoxon 검정 (비모수) ------------
# 정규성을 만족하지 않을 때 고려
# 대립가설: 무게와 120g의 차이의 중앙 위치가 0보다 작다.
stats.wilcoxon(df['무게'] - 120, alternative='less')

# ---------------------------------------------------------
# 2. 대응표본 t검정
# 키워드: 동일 대상, 전후 비교, 사전·사후, 치료 전후
# ---------------------------------------------------------

# ------------ 대응표본 t검정 ------------
# 대립가설: 교육 후 평균 점수가 더 높다.
# before - after < 0
stats.ttest_rel(df['before'], df['after'], alternative='less')

# 대립가설: 교육 후 평균 점수가 더 높다.
# after - before > 0
stats.ttest_rel(df['after'], df['before'], alternative='greater')

# ------------ Shapiro-Wilk 정규성 검정 ------------
# 대응표본은 차이값의 정규성 확인
df['diff'] = df['after'] - df['before']
stats.shapiro(df['diff'])

# ------------ Wilcoxon 검정 (비모수) ------------
# 정규성을 만족하지 않을 때 고려
# 대립가설: 교육 후 점수가 더 높다.
stats.wilcoxon(df['after'], df['before'], alternative='greater')

# ---------------------------------------------------------
# 3. 독립표본 t검정
# 키워드: 서로 다른 두 집단, 남녀, A그룹과 B그룹
# ---------------------------------------------------------

# 대립가설: 두 집단의 평균 점수는 다르다.
stats.ttest_ind(A, B)

# 등분산을 가정하지 않는 경우 (Welch t검정)
stats.ttest_ind(A, B, equal_var=False)

# 대립가설: B그룹 평균 점수가 더 높다.
# A - B < 0
stats.ttest_ind(A, B, equal_var=True, alternative='less')

# 대립가설: A그룹 평균 점수가 더 높다.
# A - B > 0
stats.ttest_ind(A, B, equal_var=True, alternative='greater')

# ------------ Shapiro-Wilk 정규성 검정 ------------
# 귀무가설: 정규분포를 따른다.
stats.shapiro(A)
stats.shapiro(B)

# ------------ Levene 등분산 검정 ------------
# 둘다 정규분호를 만족
# 귀무가설: 두 집단의 분산이 같다.
# 대립가설: 두 집단의 분산이 다르다.
stats.levene(A, B)

# 등분산을 가정하지 않는 독립표본 t검정
stats.ttest_ind(A, B, equal_var=False, alternative='less')

# ------------ Mann-Whitney U 검정 (비모수) ------------
# 정규성을 만족하지 않을 때 고려
# 대립가설: A그룹이 B그룹보다 작은 값을 가지는 경향이 있다.
stats.mannwhitneyu(A, B, alternative='less')

# ---------------------------------------------------------
# 4. 검정 결과 확인 및 해석
# ---------------------------------------------------------

# ------------ 검정통계량 및 p-value ------------
result = stats.ttest_1samp(df['무게'], 120)

print(result.statistic)  # 검정통계량
print(result.pvalue)     # p-value

# ------------ p-value 해석 ------------
# p-value < 0.05  : 귀무가설 기각
# p-value >= 0.05 : 귀무가설 기각하지 못함

# ---------------------------------------------------------
# 5. 핵심 암기
# ---------------------------------------------------------

# ------------ 단일표본 t검정 ------------
# 한 집단의 평균과 기준값 비교
stats.ttest_1samp(df['무게'], 120)

# ------------ 대응표본 t검정 ------------
# 동일 대상의 전후 평균 비교
stats.ttest_rel(df['before'], df['after'])

# ------------ 독립표본 t검정 ------------
# 서로 다른 두 집단의 평균 비교
stats.ttest_ind(A, B)

# 등분산을 가정하지 않는 경우
stats.ttest_ind(A, B, equal_var=False)

# ------------ 정규성 검정 ------------
stats.shapiro(A)

# ------------ 등분산 검정 ------------
stats.levene(A, B)

# ------------ 비모수 검정 ------------
# 단일표본
stats.wilcoxon(df['무게'] - 120)

# 대응표본
stats.wilcoxon(df['after'], df['before'])

# 독립표본
stats.mannwhitneyu(A, B)

# ------------ 대립가설 방향 ------------
# 다르다: alternative='two-sided'
# 크다  : alternative='greater'
# 작다  : alternative='less'

# ------------ 검정 결과 ------------
result = stats.ttest_ind(A, B)

print(result.statistic)  # 검정통계량
print(result.pvalue)     # p-value

# p-value < 0.05  : 귀무가설 기각
# p-value >= 0.05 : 귀무가설 기각하지 못함
```

## 범주형 데이터 분석

```python
# ---------------------------------------------------------
# 범주형 데이터 분석 - 카이제곱 검정
# ---------------------------------------------------------
from scipy import stats
import pandas as pd

# ---------------------------------------------------------
# 1. 카이제곱 적합도 검정
# ---------------------------------------------------------
# 키워드: 관측빈도, 기대빈도, 기대비율, 특정 분포와 일치하는가?
# 귀무가설: 관측빈도는 기대비율과 일치한다.
# 대립가설: 관측빈도는 기대비율과 일치하지 않는다.

# ------------ 관측빈도 / 기대빈도 ------------
# 총 300명, 기대비율 50%, 35%, 15%
ob = [150, 120, 30]
ex = [0.5 * 300, 0.35 * 300, 0.15 * 300]

# ------------ 적합도 검정 ------------
result = stats.chisquare(ob, ex)

print(result.statistic)  # 카이제곱 통계량
print(result.pvalue)     # p-value

# ---------------------------------------------------------
# 2. 카이제곱 독립성 검정
# ---------------------------------------------------------
# 키워드: 두 범주형 변수, 관련성, 연관성, 독립적인가?
# 귀무가설: 언어와 합격 여부는 독립이다.
# 대립가설: 언어와 합격 여부는 독립이 아니다.

# ------------ 방법 1. 교차표 직접 생성 ------------
# R: 합격 80명, 불합격 20명
# P: 합격 90명, 불합격 10명

df = pd.DataFrame({
    '합격': [80, 90],
    '불합격': [20, 10]
}, index=['R', 'P'])

result = stats.chi2_contingency(df)

print(result.statistic)  # 카이제곱 통계량
print(result.pvalue)     # p-value

# ------------ 방법 2. 교차표 행/열 변경 ------------
df = pd.DataFrame({
    'R': [80, 20],
    'P': [90, 10]
}, index=['합격', '불합격'])

result = stats.chi2_contingency(df)

# ------------ 방법 3. 리스트로 직접 입력 ------------
df = [[80, 20], [90, 10]]

result = stats.chi2_contingency(df)

# ------------ 방법 4. 원본 데이터에서 교차표 생성 ------------
# 원본 데이터에 '언어', '합격여부' 컬럼이 있는 경우

table = pd.crosstab(df['언어'], df['합격여부'])

result = stats.chi2_contingency(table)

print(result.statistic)  # 카이제곱 통계량
print(result.pvalue)     # p-value
print(result.dof)        # 자유도
print(result.expected_freq)  # 기대빈도

# ---------------------------------------------------------
# 3. 검정 결과 확인 및 해석
# ---------------------------------------------------------

# ------------ 검정통계량 및 p-value ------------
# result는 위에서 선택한 검정의 결과

print(result.statistic)  # 검정통계량
print(result.pvalue)     # p-value

# ------------ p-value 해석 ------------
# p-value < 0.05  : 귀무가설 기각
# p-value >= 0.05 : 귀무가설 기각하지 못함

# ---------------------------------------------------------
# 4. 핵심 암기
# ---------------------------------------------------------

# ------------ 카이제곱 적합도 검정 ------------
# 관측빈도와 기대빈도 비교
ob = [150, 120, 30]
ex = [150, 105, 45]

result = stats.chisquare(ob, ex)

# ------------ 카이제곱 독립성 검정 ------------
# 두 범주형 변수의 독립성 확인
table = pd.crosstab(df['언어'], df['합격여부'])

result = stats.chi2_contingency(table)

# ------------ 검정 결과 ------------
print(result.statistic)      # 카이제곱 통계량
print(result.pvalue)         # p-value
print(result.dof)            # 자유도 (독립성 검정)
print(result.expected_freq)  # 기대빈도 (독립성 검정)

# ------------ p-value 해석 ------------
# p-value < 0.05  : 귀무가설 기각
# p-value >= 0.05 : 귀무가설 기각하지 못함
```

## 오즈비

```python
# ---------------------------------------------------------
# 로지스틱 회귀 분석
# ---------------------------------------------------------
# 키워드: 이진 종속변수, 생존 여부, 합격 여부, 오즈비
import pandas as pd
import numpy as np
from scipy import stats
from statsmodels.formula.api import logit

# ------------ 데이터 불러오기 ------------
df = pd.read_csv(
    'https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p3/Titanic.csv'
)

#1. 
# ---------------------------------------------------------
# 1. 카이제곱 독립성 검정
#    : Gender와 Survived변수간의 독립성 검정을 실시하였을떄 카이제곱 통계량은?
# ---------------------------------------------------------
# 귀무가설: 성별과 생존 여부는 독립이다.
# 대립가설: 성별과 생존 여부는 독립이 아니다.

# ------------ 교차표 생성 ------------
table = pd.crosstab(df['Gender'], df['Survived'])
print(table)

# ------------ 독립성 검정 ------------
result = stats.chi2_contingency(table)

print(result.statistic)  # 카이제곱 통계량
print(result.pvalue)     # p-value

# ---------------------------------------------------------
# 2. 로지스틱 회귀모형
#    : Gender, SibSp, Parch, Fare를 독립변수로 사용하여 
#      로지스틱 회귀모형을 실시했을떄 parch변수의 계수값.   
# ---------------------------------------------------------
# 키워드: 종속변수가 0/1, 회귀계수, 유의확률
# 종속변수: Survived
# 독립변수: Gender, SibSp, Parch, Fare

# ------------ 데이터 확인 ------------
print(df[['Gender', 'SibSp', 'Parch', 'Fare']].head(3))

# ------------ 모델 학습 ------------
model = logit(
    'Survived ~ C(Gender) + SibSp + Parch + Fare',
    data=df
).fit()

# ------------ 회귀분석 결과 ------------
print(model.summary())

# ------------ 회귀계수 확인 ------------
print(model.params)

# Parch 변수의 회귀계수
print(model.params['Parch'])
print(round(model.params['Parch'], 3))

# ---------------------------------------------------------
# 3. 오즈비 계산
#    : 2번 문제에서 추정된 로지스틱 회귀모형에서 SibSp변수가 한 단위 증가할 때 생존할 오즈비 값은?
# ---------------------------------------------------------
# 키워드: 오즈비, Odds Ratio, Exp(B)
# 오즈비 = exp(회귀계수)

# ------------ SibSp 회귀계수 ------------
print(model.params['SibSp'])

# ------------ SibSp 오즈비 ------------
print(np.exp(model.params['SibSp']))

# 소수점 셋째 자리 반올림
print(round(np.exp(model.params['SibSp']), 3))

# ---------------------------------------------------------
# 4. 검정 결과 확인 및 해석
# ---------------------------------------------------------

# ------------ 회귀계수 및 p-value ------------
print(model.params)    # 회귀계수
print(model.pvalues)   # 변수별 p-value

# 특정 변수의 p-value
print(model.pvalues['Parch'])

# ---------------------------------------------------------
# 5. 핵심 암기
# ---------------------------------------------------------

# ------------ 카이제곱 독립성 검정 ------------
table = pd.crosstab(df['Gender'], df['Survived'])

result = stats.chi2_contingency(table)

print(result.statistic)
print(result.pvalue)

# ------------ 로지스틱 회귀모형 ------------
model = logit(
    'Survived ~ C(Gender) + SibSp + Parch + Fare',
    data=df
).fit()

print(model.summary())

# ------------ 회귀계수 ------------
print(model.params['Parch'])

# ------------ 오즈비 ------------
print(np.exp(model.params['SibSp']))

# ------------ p-value ------------
print(model.pvalues['Parch'])

# ------------ 핵심 구분 ------------
# C(변수명)                  : 범주형 변수 처리
# model.params['변수명']     : 회귀계수
# np.exp(model.params['변수명']) : 오즈비
# model.pvalues['변수명']    : p-value

```

## 회귀분석

```python
# ---------------------------------------------------------
# 회귀 분석
# ---------------------------------------------------------
import pandas as pd
from scipy import stats
from statsmodels.formula.api import ols
from sklearn.metrics import mean_squared_error

# ---------------------------------------------------------
# 1. 상관관계 분석
# ---------------------------------------------------------
# 키워드: 상관관계, 상관계수, 선형 관계, 유의확률

# ------------ 상관계수 ------------
# 기본값: 피어슨 상관계수
df.corr(numeric_only=True)

# 두 변수의 상관계수
df['키'].corr(df['몸무게'])
df['몸무게'].corr(df['키'])

# ------------ 피어슨 상관계수 ------------
# 두 수치형 변수의 선형 상관관계
df.corr(method='pearson', numeric_only=True)

# ------------ 스피어만 상관계수 ------------
# 순위 기반 상관관계
df.corr(method='spearman', numeric_only=True)

# ------------ 켄달타우 상관계수 ------------
# 순위 기반 상관관계
df.corr(method='kendall', numeric_only=True)

# ------------ 피어슨 상관관계 검정 ------------
# 귀무가설: 두 변수의 모집단 상관계수는 0이다.
# 대립가설: 두 변수의 모집단 상관계수는 0이 아니다.
result = stats.pearsonr(df['몸무게'], df['키'])

print(result.statistic)  # 상관계수
print(result.pvalue)     # p-value

# ------------ 스피어만 상관관계 검정 ------------
result = stats.spearmanr(df['몸무게'], df['키'])

print(result.statistic)
print(result.pvalue)

# ------------ 켄달타우 상관관계 검정 ------------
result = stats.kendalltau(df['몸무게'], df['키'])

print(result.statistic)
print(result.pvalue)

# ---------------------------------------------------------
# 2. 단순선형회귀
# ---------------------------------------------------------
# 키워드: 독립변수 1개, 회귀계수, 결정계수, 예측
# 종속변수: 키
# 독립변수: 몸무게

# ------------ 모델 학습 ------------
model = ols('키 ~ 몸무게', data=df).fit()

# ------------ 회귀분석 결과 ------------
print(model.summary())

# ------------ 결정계수 (R-squared) ------------
print(model.rsquared)

# ------------ 회귀계수 ------------
# 기울기
print(model.params['몸무게'])

# 절편
print(model.params['Intercept'])

# ------------ p-value ------------
# 귀무가설: 몸무게의 회귀계수는 0이다.
# 대립가설: 몸무게의 회귀계수는 0이 아니다.
print(model.pvalues['몸무게'])

# ------------ 예측값 ------------
# 몸무게가 50일 때 키 예측
newdata = pd.DataFrame({
    '몸무게': [50]
})

print(model.predict(newdata))

# ------------ 잔차 제곱합 (SSE) ------------
# 잔차 = 실제값 - 예측값
df['잔차'] = df['키'] - model.predict(df)

print((df['잔차'] ** 2).sum())

# 간단한 방법
print((model.resid ** 2).sum())

# ------------ 평균제곱오차 (MSE) ------------
print((model.resid ** 2).mean())

# 사이킷런 MSE
pred = model.predict(df)
print(mean_squared_error(df['키'], pred))

# ------------ 회귀계수 신뢰구간 ------------
# 95% 신뢰구간
print(model.conf_int(alpha=0.05))

# ------------ 예측값 신뢰구간 및 예측구간 ------------
# 몸무게가 50일 때
newdata = pd.DataFrame({
    '몸무게': [50]
})

pred = model.get_prediction(newdata)
print(pred.summary_frame(alpha=0.05))

# ---------------------------------------------------------
# 3. 다중선형회귀
# ---------------------------------------------------------
# 키워드: 독립변수 2개 이상, 회귀계수, 결정계수
# 종속변수: 매출액
# 독립변수: 광고비, 플랫폼

# ------------ 모델 학습 ------------
model = ols('매출액 ~ 광고비 + 플랫폼', data=df).fit()

# ------------ 회귀분석 결과 ------------
print(model.summary())

# ------------ 결정계수 (R-squared) ------------
print(model.rsquared)

# ------------ 회귀계수 ------------
# 광고비
print(model.params['광고비'])

# 플랫폼
print(model.params['플랫폼'])

# 절편
print(model.params['Intercept'])

# ------------ p-value ------------
# 귀무가설: 해당 변수의 회귀계수는 0이다.
# 대립가설: 해당 변수의 회귀계수는 0이 아니다.

print(model.pvalues['광고비'])
print(model.pvalues['플랫폼'])

# ------------ 예측값 ------------
# 광고비 50, 플랫폼 20일 때 매출액 예측
newdata = pd.DataFrame({
    '광고비': [50],
    '플랫폼': [20]
})

print(model.predict(newdata))

# ------------ 잔차 제곱합 (SSE) ------------
print((model.resid ** 2).sum())

# ------------ 평균제곱오차 (MSE) ------------
print((model.resid ** 2).mean())

# ------------ 회귀계수 신뢰구간 ------------
# 광고비, 플랫폼 회귀계수의 95% 신뢰구간
print(model.conf_int(alpha=0.05))

# 특정 변수의 신뢰구간
print(model.conf_int().loc['광고비'])
print(model.conf_int().loc['플랫폼'])

# ------------ 예측값 신뢰구간 및 예측구간 ------------
newdata = pd.DataFrame({
    '광고비': [50],
    '플랫폼': [20]
})

pred = model.get_prediction(newdata)
print(pred.summary_frame(alpha=0.05))

# ---------------------------------------------------------
# 4. 다중선형회귀 (범주형 변수)
# ---------------------------------------------------------
# 종속변수: 매출액
# 독립변수: 광고비, 유형
# 유형: A, B, C (범주형 변수)

# ------------ 방법 1. C() 사용 (추천) ------------
# 범주형 변수를 자동으로 더미변수 처리
model = ols('매출액 ~ 광고비 + C(유형)', data=df).fit()

print(model.summary())
print(model.params)
print(model.pvalues)

# ------------ 방법 2. 원핫인코딩 ------------
# 범주형 변수를 직접 더미변수로 변환
df2 = pd.get_dummies(df, columns=['유형'], drop_first=True, dtype=int)

print(df2.head())

# 회귀 모델 학습
model = ols(
    '매출액 ~ 광고비 + 유형_B + 유형_C',
    data=df2
).fit()

print(model.summary())

# ---------------------------------------------------------
# 5. 검정 결과 확인 및 해석
# ---------------------------------------------------------

# ------------ 주요 결과 ------------
print(model.rsquared)    # 결정계수
print(model.params)      # 회귀계수
print(model.pvalues)     # p-value
print(model.conf_int())  # 회귀계수 신뢰구간

# ------------ p-value 해석 ------------
# p-value < 0.05  : 귀무가설 기각
# p-value >= 0.05 : 귀무가설 기각하지 못함

# ---------------------------------------------------------
# ** 핵심 암기 코드 **
# ---------------------------------------------------------

# ------------ 상관관계 ------------
from scipy import stats

result = stats.pearsonr(df['키'], df['몸무게'])
print(result.statistic)
print(result.pvalue)

# ------------ 선형회귀 ------------
from statsmodels.formula.api import ols

model = ols('매출액 ~ 광고비 + 플랫폼', data=df).fit()

print(model.summary())
print(model.rsquared)
print(model.params)
print(model.pvalues)

# ------------ 잔차 제곱합 / MSE ------------
print((model.resid ** 2).sum())
print((model.resid ** 2).mean())

# ------------ 회귀계수 신뢰구간 ------------
print(model.conf_int())

# ------------ 예측값 / 신뢰구간 / 예측구간 ------------
newdata = pd.DataFrame({
    '광고비': [50],
    '플랫폼': [20]
})

print(model.predict(newdata))
print(model.get_prediction(newdata).summary_frame(alpha=0.05))

```

## 분산분석(ANOVA)

```python
# ---------------------------------------------------------
# 분산 분석
# ---------------------------------------------------------
import pandas as pd
from scipy import stats
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
from statsmodels.stats.multicomp import pairwise_tukeyhsd, MultiComparison

# ---------------------------------------------------------
# 1. 일원 분산 분석 (One-way ANOVA)
# ---------------------------------------------------------
# 키워드: 세 집단 이상, 평균 비교, 하나의 요인
# 귀무가설: 모든 집단의 평균은 같다.
# 대립가설: 적어도 한 집단의 평균은 다르다.

# ------------ 일원 분산 분석 ------------
result = stats.f_oneway(df['A'], df['B'], df['C'], df['D'])

print(result.statistic)  # F통계량
print(result.pvalue)     # p-value

# ------------ Shapiro-Wilk 정규성 검정 ------------
# 귀무가설: 정규분포를 따른다.
# 대립가설: 정규분포를 따르지 않는다.
print(stats.shapiro(df['A']))
print(stats.shapiro(df['B']))
print(stats.shapiro(df['C']))
print(stats.shapiro(df['D']))

# ------------ Levene 등분산 검정 ------------
# 귀무가설: 모든 집단의 분산은 같다.
# 대립가설: 적어도 한 집단의 분산은 다르다.
print(stats.levene(df['A'], df['B'], df['C'], df['D']))

# ---------------------------------------------------------
# 2. 일원 분산 분석 - 심화
# ---------------------------------------------------------
# 키워드: ANOVA 테이블, 사후검정, 비모수 검정

# ------------ 데이터 재구조화 ------------
# 넓은 형태 → 긴 형태
df_melt = df[['A', 'B', 'C', 'D']].melt()
df_melt = df[['A', 'B', 'C', 'D']].melt().dropna() # 결측치 있을떄
print(df_melt.head())

# ------------ ANOVA 테이블 ------------
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
model = ols('value ~ C(variable)', data=df_melt).fit()
print(anova_lm(model))

# ------------ 사후검정 ------------
# 어떤 그룹들 간에 통꼐적으로 유의미한 차이가 있는지 구체적으로 파악하는 것 
from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd, MultiComparison

# ------------ Tukey HSD 사후검정 ------------
# 어떤 집단 간 평균 차이가 있는지 확인
tukey_result = pairwise_tukeyhsd(
    df_melt['value'],
    df_melt['variable'],
    alpha=0.05
)

print(tukey_result.summary())

# ------------ Bonferroni 사후검정 ------------
# 여러 집단을 쌍별로 비교하고 p-value 보정
mc = MultiComparison(df_melt['value'], df_melt['variable'])

bon_result = mc.allpairtest(
    stats.ttest_ind,
    method='bonf'
)

print(bon_result[0])

# ------------ Kruskal-Wallis 검정 (비모수) ------------
# 정규성을 만족하지 않을 때 고려
# 귀무가설: 집단들의 분포가 같다.
# 대립가설: 적어도 한 집단의 분포가 다르다.

result = stats.kruskal(df['A'], df['B'], df['C'], df['D'])

print(result.statistic)
print(result.pvalue)

# ---------------------------------------------------------
# 3. 이원 분산 분석 (Two-way ANOVA)
# ---------------------------------------------------------
# 키워드: 두 개의 요인, 주효과, 상호작용
# 종속변수: 토마토수
# 독립변수: 종자, 비료

# ------------ 이원 분산 분석 ------------
# C(): 범주형 변수 처리
# * : 주효과 + 상호작용 포함

model = ols(
    '토마토수 ~ C(종자) * C(비료)',
    data=df
).fit()

# ------------ ANOVA 테이블 ------------
print(anova_lm(model))

# ------------ 주효과 및 상호작용 ------------
# C(종자)          : 종자에 따른 평균 차이
# C(비료)          : 비료에 따른 평균 차이
# C(종자):C(비료)  : 종자와 비료의 상호작용

# ------------ 다른 작성 방법 (동일한 모형) ------------
model = ols(
    '토마토수 ~ C(종자) + C(비료) + C(종자):C(비료)',
    data=df
).fit()

print(anova_lm(model))

# ---------------------------------------------------------
# 4. 이원 분산 분석 - 심화
# ---------------------------------------------------------
# 키워드: 사후검정, 집단별 평균 차이

# ------------ Tukey HSD 사후검정 ------------
# 종자별 비교
tukey_result = pairwise_tukeyhsd(
    df['토마토수'],
    df['종자'],
    alpha=0.05
)

print(tukey_result.summary())

# ------------ Tukey HSD 사후검정 ------------
# 비료별 비교
tukey_result = pairwise_tukeyhsd(
    df['토마토수'],
    df['비료'],
    alpha=0.05
)

print(tukey_result.summary())

# ------------ Bonferroni 사후검정 ------------
# 종자별 비교
mc = MultiComparison(df['토마토수'], df['종자'])

bon_result = mc.allpairtest(
    stats.ttest_ind,
    method='bonf'
)

print(bon_result[0])

# ------------ Bonferroni 사후검정 ------------
# 비료별 비교
mc = MultiComparison(df['토마토수'], df['비료'])

bon_result = mc.allpairtest(
    stats.ttest_ind,
    method='bonf'
)

print(bon_result[0])

# ---------------------------------------------------------
# 5. 정규성 및 등분산 검정
# ---------------------------------------------------------
# 이원 분산분석에서 각 요인 조합별 확인

# ------------ Shapiro-Wilk 정규성 검정 ------------
# 각 종자·비료 조합의 정규성 확인
for (seed, fertilizer), group in df.groupby(['종자', '비료']):
    print(seed, fertilizer, stats.shapiro(group['토마토수']))

# ------------ Levene 등분산 검정 ------------
# 종자별 등분산 검정
groups = [group['토마토수'] for _, group in df.groupby('종자')]
print(stats.levene(*groups))

# 비료별 등분산 검정
groups = [group['토마토수'] for _, group in df.groupby('비료')]
print(stats.levene(*groups))

# 종자·비료 조합별 등분산 검정
groups = [
    group['토마토수']
    for _, group in df.groupby(['종자', '비료'])
]
print(stats.levene(*groups))

# ---------------------------------------------------------
# 6. 검정 결과 확인 및 해석
# ---------------------------------------------------------

# ------------ 일원 분산 분석 ------------
result = stats.f_oneway(df['A'], df['B'], df['C'], df['D'])

print(result.statistic)  # F통계량
print(result.pvalue)     # p-value

# ------------ p-value 해석 ------------
# p-value < 0.05  : 귀무가설 기각
# p-value >= 0.05 : 귀무가설 기각하지 못함

# ---------------------------------------------------------
# ** 핵심 암기 코드 **
# ---------------------------------------------------------

# ---------------------------------------------------------
# 1. 일원 분산 분석
# ---------------------------------------------------------
from scipy import stats

# 정규성
print(stats.shapiro(df['A']))

# 등분산
print(stats.levene(df['A'], df['B'], df['C']))

# 일원 분산 분석
result = stats.f_oneway(df['A'], df['B'], df['C'])
print(result.statistic)
print(result.pvalue)

# ---------------------------------------------------------
# 2. 이원 분산 분석
# ---------------------------------------------------------
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm

model = ols('토마토수 ~ C(종자) * C(비료)', data=df).fit()

print(anova_lm(model))

# ---------------------------------------------------------
# 3. 사후검정
# ---------------------------------------------------------
from statsmodels.stats.multicomp import pairwise_tukeyhsd

result = pairwise_tukeyhsd(
    df_melt['value'],
    df_melt['variable'],
    alpha=0.05
)

print(result.summary())

# ---------------------------------------------------------
# 4. 비모수 검정
# ---------------------------------------------------------
result = stats.kruskal(df['A'], df['B'], df['C'])

print(result.statistic)
print(result.pvalue)

```
