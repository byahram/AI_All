# Folium

- 지도 시각화에 유용한 도구

- 세계 지도를 기본 지원하고 다양한 스타일의 지도 이미지를 제공

## 지도 만들기

- Map() 함수를 이용하여 지도 객체 생성

- 지도 화면은 줌 과 이동 기능을 지원

- folium은 웹 기반 지도를 만들어 웹 기반 IDE 에서는 지도 객체를 바로 확인할 수 있고, 웹 기반 IDE가 아니라면 HTML 파일로 저장하고 확인

```python
import pandas as pd
import folium
import json
```

```python
# 서울 지도 만들기
seoul_map = folium.Map(location = [37.55, 126.98], zoom_start = 12)
seoul_map
```

```python
# 지도를 html 파일로 저장하기
seoul_map.save("seoul.html")
```

## 지도 스타일 적용

- Map() 함수에 tiles 옵션을 적용하여 스타일을 변경할 수 있음

```python
seoul_map2 = folium.Map(location = [37.55, 126.98], tiles = "Cartodb Positron", zoom_start = 12)
seoul_map3 = folium.Map(location = [37.55, 126.98], tiles = "Cartodb dark_matter", zoom_start = 15)
```

```python
seoul_map2
```

```python
seoul_map3
```

## 지도에 마커 표시하기

```python
df = pd.read_excel("./data/서울지역 대학교 위치.xlsx", index_col = 0)
df.head()
```

```python
# 대학교 위치 정보를 Marker로 표시
for idx, row in df.iterrows():
    folium.Marker([row["위도"], row["경도"]], popup = idx).add_to(seoul_map)
```

```python
seoul_map
```

## 지도에 원형 마커 표시

```python
seoul_map = folium.Map(location = [37.55, 126.98], zoom_start = 12)
```

```python
for idx, row in df.iterrows():
    folium.CircleMarker([row["위도"], row["경도"]],
                        radius = 10, # 원의 반지름
                        color = "brown", # 원의 둘레 색상
                        fill = True,
                        fill_color = "cora", # 원을 채우는 색
                        fill_opacity = 0.7, # 투명도
                        popup = idx).add_to(seoul_map)
```

```python
seoul_map
```

## 지도 영역에 단계구분도(Choropleth Map) 표시하기

```python
df = pd.read_excel("./data/경기도인구데이터.xlsx", index_col = "구분")
df.head()
```

```python
df.columns
```

```python
df.columns = df.columns.map(str)
```

```python
df.columns
```

```python
geo_data = json.load(open("./data/경기도행정구역경계.json", encoding = "utf-8"))
```

```python
geo_data
```

```python
g_map = folium.Map(location = [37.5502, 126.982], stiles = "Cartodb Positron", zoom_start = 9)
```

```python
df.head()
```

```python
df.index
```

```python
folium.Choropleth(geo_data = geo_data, # 지도 경계
                  data = df["2017"], # 표시하려는 데이터
                  fill_color = "YlOrRd", fill_opacity = 0.7, line_opacity = 0.3,
                  threshold_scale = [10000, 100000, 300000, 500000, 700000],
                  key_on = "feature.properties.name").add_to(g_map)
```
