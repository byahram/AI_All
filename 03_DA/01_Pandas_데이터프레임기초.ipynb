## 행단위 데이터 추출

- loc
    - 인덱스를 기준으로 행 데이터 추출
 
- iloc
    - 행 번호를 기준으로 행 데이터 추출

### loc

- 인덱스(index) : 값의 위치를 나타낸 값
    - 데이터프레임을 만들면 자동으로 인덱스 번호가 부여됨
        - 현재 gapminder 데이터의 왼쪽에 세로로 나열된 0, 1, 2, ...

```python
df.loc[0]
```

```python
df.loc[99]
```

```python
# ord() : 문자를 전달하면 해당 문자의 유니코드 값을 반환
# chr() : 특정 문자의 유니코드 값을 전달하면 해당 문자를 반환

print(ord("a"), ord("z"))
```

```python
for i in range(97, 123):
    print(chr(i), end = " ")
```

```python
df_index = pd.DataFrame({"value" : [i for i in range(26)]},
                        index = [chr(i) for i in range(97, 123)])
```

```python
df_index.head()
```

```python
df_index.loc["a"]
```

```python
df_index.loc["z"]
```

```python
# 여러 행 추출
df_index.loc[["a", "c"]]
```

## 시리즈 연산

```python
scientists = pd.read_csv("./data/scientists.csv")
```

```python
scientists.head()
```

### 시리즈의 기초통계 메서드

```python
# 평균
scientists["Age"].mean()
```

```python
# 최솟값
scientists["Age"].min()
```

```python
# 최댓값
scientists["Age"].max()
```

```python
# 표준편차
scientists["Age"].std()
```

### 필터링

```python
ages = scientists["Age"]
```

```python
print(ages)
```

```python
# 평균보다 나이가 많은 사람의 데이터 추출
ages[ages > ages.mean()]
```

```python
ages > ages.mean()
```

```python
ages[[True, True, False, False, True, True, False, True]]
```

### 시리즈의 사칙연산

```python
# 시리즈와 시리즈의 사칙연산
ages + ages
```

```python
ages * ages
```

```python
# 시리즈와 숫자 연산
ages + 100
```

```python
ages * 2
```

```python
# 길이가 서로 다른 시리즈를 연산하면 인덱스가 일치한 데이터만 계산
ages + pd.Series([1, 100])
```

```python
# 인덱스에 따라 내림차순 정렬
ages.sort_index(ascending = False)
```

```python
# 눈에 보이는 순서대로 연산하는 것이 아닌 인덱스가 일치하는 값끼리 연산
ages + ages.sort_index(ascending = False)
```

## datetime

```python
scientists.dtypes
```

```python
pd.to_datetime(scientists["Born"], format = "%Y-%m-%d")
```

```python
scientists["born_dt"] = pd.to_datetime(scientists["Born"], format = "%Y-%m-%d")
scientists["died_dt"] = pd.to_datetime(scientists["Died"], format = "%Y-%m-%d")
```

```python
scientists.head()
```

```python
scientists.dtypes
```

```python
scientists["died_dt"] - scientists["born_dt"]
```
