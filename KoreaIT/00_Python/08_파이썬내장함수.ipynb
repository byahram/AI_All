## 문자열 내장 함수

### 1. chr() / ord()

- 모든 문자는 각각 문자 코드라고 하는 자신만의 코드 값이 있음
- chr() : 특정 문자의 유니코드 값을 전달하면 해당 문자를 반환
- ord() : 문자를 전달하면 해당 문자의 유니코드 값을 반환

````python
ord("0")
````

````python
ord("1")
````

````python
chr(48)
````

````python
chr(49)
````

````python
ord("a"), ord("z")
````

````python
for i in range(97, 123):
    print(chr(i), end = " ")
````

````python
"100 + 200"
````

````python
a = 10

eval("a * 5")
````

````python
eval("min(1, 2, 3)")
````

````python
eval("[1, 2, 3, 4, 5]")
````

## 숫자 내장 함수

### 1. abs()

- 절댓값을 반환

````python
abs(20)
````

````python
abs(-10)
````

### 2. divmod()

- 전달된 두 인수를 나누어 몫과 나머지를 반환

````python
divmod(10, 3)
````

````python
money = 10000
price = 3000

n, remain = divmod(money, price)
print(f"빵을 {n}개 사고 {remain}원이 남았습니다")
````

### 3. max() / min()

- 전달된 인수 중 최댓값을 반환 / 최솟값을 반환

````python
max(1, 10)
````

````python
min(1, 10)
````

````python
lst = [1, 2, 3, 4, 5]

print(max(lst), min(lst))
````

## 시퀀스 내장 함수

### 1. enumerate()

- 리스트에 저장된 요소와 인덱스가 함께 추출

````python
for idx, element in enumerate(["가위", "바위", "보"]):
    print(idx, element)
````

### 2. range()

- 특정 범위의 값을 생성

- 주로 for문과 함께 사용

- list() 함수로 리스트로 변환해서 사용하기도 함

```
range(stop)
range(start, stop)
range(start, stop, step)
start <= n < stop
```

````python
list(range(10))
````

````python
list(range(1, 11))
````

````python
list(range(0, 30, 5))
````

````python
list(range(0, -10, -1))
````

````python
list(range(0))
````

````python
list(range(1, 0))
````

### 3. len()

- 객체의 길이(항목의 수)를 반환

````python
lst = ["a", "b", "c", "d"]

print(len(lst))
````

````python
dic = {"a" : "apple", "b" : "banana"}

print(len(dic))
````

````python
len(range(10))
````

````python
len(range(1, 10))
````

### 4. sorted()

- 오름차순, 내림차순 정렬 결과를 반환

````python
sorted(my_list, reverse = True)
````

### 5. zip()

- 여러 개의 반복가능객체의 각 요소를 묶어서 반환

- 전달된 객체들의 길이가 다르면 짧은 객체를 기준으로 동작

````python
names = ["james", "emily", "amanda"]
scores = [60, 70, 80]

for name, score in zip(names, scores):
    print(f"{name}의 점수는 {score}점 입니다.")
````
