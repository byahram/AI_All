- html파일을 열어보면 아무것도 없는 깨끗한 페이지가 나옴

- 이 상태가 html의 가장 기초적인 상태

```python
# index.html
<!DOCTYPE html>
<html>
    <head> <!-- head : 문서에 대한 기본적인 정보가 담기는 태그 -->
        <title>전준형의 홈페이지</title> <!-- title : 브라우저 위(탭)에 나타나는 제목 -->
    </head>
    <body> <!-- body : 주로 화면에 나타나는 내용을 담음 -->
        <nav> <!-- nav : 웹사이트의 주요 페이지를 안내하는 네비게이션 역할을 만들 때 사용 -->
            <!-- a : 해당 부분을 클릭하면 페이지가 이동하도록 만들고 싶을 때 사용 -->
            <a href="./index.html">Home</a>
            <a href="./blog_list.html">Blog</a> <!-- 넘어갈 링크는 href 로 지정 -->
            <a href="./about_me.html">About me</a>
        </nav>

        <h1>첫 번째 크기 헤드라인</h1> <!-- h1 ~ h5 : 헤드라인 -->
        <h2>두 번째 크기 헤드라인</h2>
        <h3>세 번째 크기 헤드라인</h3>
        <h4>네 번째 크기 헤드라인</h4>
        <h5>다섯 번째 크기 헤드라인</h5>

        <p>문단은 p로 쓰기. Paragraph의 앞 글자를 따온 것</p>
        <a href="https://google.com">Go to google</a>
        <hr/> <!-- hr/ : 수평선을 그어주는 태그 -->
        <img src = "https://wallpapercave.com/wp/wp15004701.webp" width = "600">
    </body>
</html>
```

- 스타일 적용하기
    - style 속성을 이용해 배경색, 글꼴 등을 바꿔줄 수도 있음

```python
# about_me.html

    <body>
        <!-- style 속성을 추가하여 배경색은 darkgreen으로 지정, 글자크기 150%, 텍스트 가운데 정렬 -->
        <nav style="background-color: darkgreen; font-size: 150%; text-align: center">
            <a href="./index.html" style="color: white">Home</a>
            <a href="./blog_list.html" style="color: white">Blog</a>
            <a href="./about_me.html" style="color: white">About me</a>
        </nav>
```

- 이렇게 페이지를 만든 이후에 nav 태그 안 a태그의 글자 색상을 다른 색으로 수정하고 싶다면
    - 하나하나 일일이 수정해야하는데 번거로운 일임
    - 따라서 head태그 안에 style 태그를 삽입하고, 그 안에 원하는 태그 스타일을 지정해서 해결할 수 있음

```python
# about_me.html

<!DOCTYPE html>
<html>
    <head>
        <title>About me</title>
        <style>
            nav {
                background-color: darkgreen;
                font-size: 150%;
                text-align: center;
            }
            nav a {
                color: gold;
            }
        </style><!-- style 태그 안에서 nav태그의 스타일과 nav 태그 안 a태그의 스타일도 설정 -->
    </head>

    <body>
        <!-- style 속성을 추가하여 배경색은 darkgreen으로 지정, 글자크기 150%, 텍스트 가운데 정렬 -->
        <nav>
            <a href="./index.html">Home</a>
            <a href="./blog_list.html">Blog</a>
            <a href="./about_me.html">About me</a>
        </nav>
```

# 자바스크립트

- 사용자가 서버에 특정 페이지를 요청하면 서버는 html파일과 css파일을 넘겨주고, 사용자의 컴퓨터나 스마트폰의 브라우저가 이 파일을 렌더링해서 화면에 보여주는 방식

- 이 방법은 페이지 내용 중 작은 요소라도 바뀌면 서버에 다시 요청해서 새 html 파일과 css파일 전체를 다시 받아와야함

- 사용자가 각각 숫자를 입력하고 실시간으로 그 덧셈을 계산해 화면에 출력하는 기능을 웹페이지에 추가하려면
    - 웹 브라우저 안에 동적인 부분을 만들 땐 자바스크립트(javascript) 라는 프로그래밍 언어를 사용

```python
# index.html

<!DOCTYPE html>
<html>
    <head> <!-- head : 문서에 대한 기본적인 정보가 담기는 태그 -->
        <title>전준형의 홈페이지</title> <!-- title : 브라우저 위(탭)에 나타나는 제목 -->
        <link href="./practice.css" rel="stylesheet" type="text/css">

        <script type="text/javascript">
            function doSomething() {
                let a = document.getElementById('inputA').value;
                let b = document.getElementById('inputB').value;
                document.getElementById("valueA").innerHTML = a;
                document.getElementById("valueB").innerHTML = b;
                document.getElementById("valueC").innerHTML = Number(a) + Number(b);
            }
        </script>
    </head>
```

- 자바스크립트 코드는 script 태그 안에 담고

- 그 안에 자바스크립트 함수 doSomething() 이 선언됨

- 함수가 하는 일
    - 문서 안의 요소 중 id가 inputA, inputB인 input 태그의 요소값을 가져와 변수 a, b에 저장
    - id가 valueA, valueB인 요소의 텍스트를 a, b 값으로 수정
    - id가 valueC인 요소의 텍스트는 a와 b의 값을 숫자로 바꾼 다음 더한 결괏값으로 수정
 
- 자바스크립트 함수에 활용할 HTML 태그 추가
    - doSomething() 함수에서 사용할 inputA와 inputB라는 id로 input 태그를 추가해야함

```python
# index.html

        <h5>다섯 번째 크기 헤드라인</h5>

        <label for="inputA">a</label>
        <input id="inputA" value="1" onkeyup="doSomething()">
        <label for="inputB">b</label>
        <input id="inputB" value="2" onkeyup="doSomething()">

        <p><span id="valueA">1</span> + <span id="valueB">2</span> = <span id="valueC">3</span>입니다.</p>

        <p>문단은 p로 쓰기. Paragraph의 앞 글자를 따온 것</p>
```

- 현재 시간 확인 기능 추가

- whatTimeIsIt() 함수 추가

```python
        <script type="text/javascript">
            function doSomething() {
                let a = document.getElementById('inputA').value;
                let b = document.getElementById('inputB').value;
                document.getElementById("valueA").innerHTML = a;
                document.getElementById("valueB").innerHTML = b;
                document.getElementById("valueC").innerHTML = Number(a) + Number(b);
            }

            function whatTimeIsIt() {
                alert(new Date());
            }
        </script>
```

```python
# index.html

        <h1>첫 번째 크기 헤드라인</h1> <!-- h1 ~ h5 : 헤드라인 -->
        <h2>두 번째 크기 헤드라인</h2>
        <h3>세 번째 크기 헤드라인</h3>
        <h4>네 번째 크기 헤드라인</h4>
        <h5>다섯 번째 크기 헤드라인</h5>

        <button onclick="whatTimeIsIt()">현재시간</button>
        <hr/>
```

- js파일을 여러 html파일에 적용하기
    - 다른 html파일에서도 이 자바스크립트 기능을 사용하고 싶다면 파일 하나하나 코드 복붙하려면 번거로움
    - 앞에서 css 파일을 만들어서 적용했듯이 js파일을 만들어서 html에 적용

```python
# add_two_numbers.js

function doSomething() {
    let a = document.getElementById('inputA').value;
    let b = document.getElementById('inputB').value;
    document.getElementById("valueA").innerHTML = a;
    document.getElementById("valueB").innerHTML = b;
    document.getElementById("valueC").innerHTML = Number(a) + Number(b);
}
```

```python
# what_time_is_it.js

function whatTimeIsIt() {
    alert(new Date());
}
```
