## HTML(Hyper Text Markup Language)

- 웹페이지 표시를 위해 개발된 마크업 언어

- HTML 문서는 태그들로 이루어져 있으며, 태그는 두 가지 형태를 가짐
    - 열고 닫는 형태 : 태그 사이에 다른 태그나 내용을 넣을 수 있음
        - 예) \<div>내용\</div>
 
    - 단독 형태 : 태그 사이에 내용을 넣을 수 없으며, 단독으로 쓰임
        - 예) \<img src = "이미지주소">

- 자주 사용하게 되는 태그들
    - div 태그 : 특별한 의미를 갖지 않으며, 문서의 영역을 수평으로 나눌 때(division) 사용
 
    - p 태그 : 문장(paragraph)을 나타낼 때 사용
 
    - img 태그 : 이미지를 첨부할 때 사용
 
    - ol, ul, li 태그 : 리스트(ordered list, unordered list)와 리스트 아이템(list item)을 나타낼 때 사용
 
    - h1, h2, h3, h4 태그 : 헤더(header)를 나타내며 숫자가 작을수록 상위 항목임

### 가져온 데이터를 Template으로 전달

- view 함수는 Model 클래스를 사용해 데이터베이스로부터 원하는 데이터를 가져오고, 그 데이터를 템플릿으로 전달해주는 역할을 함

- Template은 view함수가 전달해준 데이터를 사용해서 동적으로 HTML을 구성
    - 동적 구성 : 데이터베이스의 데이터에 따라 다른 HTML을 그때그때 만들어서 보여준다는 의미
 
- Template에 데이터를 전달할 때는 파이썬 딕셔너리 객체를 사용해 전달
    - 관용적으로 Template에 전달하는 딕셔너리 객체의 변수명은 context를 사용

```python
# pyburger/views.py

def burger_list(request):
    burgers = Burger.objects.all()
    print("전체 햄버거 목록 :", burgers)

    # Template으로 전달해줄 dict 객체
    context = {
        "burgers" : burgers, # burgers 키에 burgers 변수(QuerySet객체)를 전달
    }
    return render(request, "burger_list.html", context)
```

## GET을 이용한 버거 검색

### view에 데이터 전달

- 쿼리스트링을 이용하면 get방식으로 서버에 데이터를 전송할 수 있음

- 전달된 데이터는 view에서도 확인 가능
    - 전달된 값은 view 함수의 첫 번째 매개변수인 request의 GET속성으로 전달됨

```python
# pyburger/views.py

def burger_search(request):
    print(request.GET) # request.GET으로 전달된 데이터를 출력
    return render(request, "burger_search.html")
```

- QueryDict 객체로 데이터가 전달되는데, 파이썬의 딕셔너리와 유사한 객체

```python
# pyburger/views.py

def burger_search(request):
    keyword = request.GET["keyword"]
    print(keyword)
    print(request.GET) # request.GET으로 전달된 데이터를 출력

    # 이름(name 속성)에 전달받은 키워드 값이 포함된 버거를 검색
    burgers = Burger.objects.filter(name__contains = keyword)
    print(burgers)
    return render(request, "burger_search.html")
```

- filter 함수는 조건과 일치하는 객체를 모두 반환
    - name__contains = keyword 는 name 속성이 keyword 변수의 값을 포함하는 경우를 뜻함
 
- 테스트해보면 서버 로그에는 버거 목록이 나오지만 브라우저에는 검색 결과가 출력되지 않음
    - 검색결과를 template에 전달하고, template에서 해당 내용을 표시해서 브라우저에 검색 결과가 나오도록 해야함

```python
# pyburger/views.py

def burger_search(request):
    keyword = request.GET["keyword"]
    print(keyword)
    print(request.GET) # request.GET으로 전달된 데이터를 출력

    # 이름(name 속성)에 전달받은 키워드 값이 포함된 버거를 검색
    burgers = Burger.objects.filter(name__contains = keyword)
    print(burgers)
    context = {
        "burgers" : burgers
    }
    return render(request, "burger_search.html", context)
```

```python
# templates/burger_search.html

<!DOCTYPE html>
<html lang="ko">
    <body>
        <h1>pyburger의 햄버거 검색화면입니다</h1>
        <h2>검색 결과</h2>
        {% for burger in burgers %}
            <div>
                {{ burger.name }}
                (가격 : {{ burger.price }}원,
                칼로리 : {{ burger.calories }}kcal)
            </div>
        {% endfor %}
    </body>
</html>
```

- 검색할 키워드에 해당하는 이름을 가진 버거 목록을 보여주는 데 성공했지만

- 검색어를 입력하기위해 주소표시줄에 쿼리스트링을 매번 직접 입력하는 것은 번거로움
    - 검색창이 필요

```python
# keyword 쿼리스트링을 빼고 접속했을 때 에러 발생하는 현상 해결

def burger_search(request):
    keyword = request.GET.get("keyword", None)
    print(keyword)
    print(request.GET) # request.GET으로 전달된 데이터를 출력

    if keyword: # keyword 값이 주어진 경우
        # 이름(name 속성)에 전달받은 키워드 값이 포함된 버거를 검색
        burgers = Burger.objects.filter(name__contains = keyword)

    else: # keyword가 주어지지 않아, None이 할당된 경우
        burgers = Burger.objects.none() # 빈 QuerySet을 할당

    context = {
        "burgers" : burgers
    }
    return render(request, "burger_search.html", context)
```

- 브라우저가 사용자가 입력한 데이터를 처리할 수 있도록 input 태그와 button 태그의 속성을 적절히 설정하고 해당 요소들을 form 태그 내부에 위치하게 해야 검색 기능이 실행됨
